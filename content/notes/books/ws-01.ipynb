```python
class Car:
    def __init__(self, brand, model, year, speed):
        self.brand = brand
        self.model = model
        self.year = year
        self.speed = speed
        self.acumulado = [self.speed]

    def honk(self):
        print("Beep beep!")

    def accelerate(self, rate):
        self.speed += rate
        # 3. Impresión de cambio diferencial de velocidad
        print(f"{self.speed=}")
        # 4.
        self.acumulado.append(self.speed)

    def get_speed(self):
        return self.speed

    # 1. Creación método de frenado
    def freno(self):
        self.speed /= 2
        self.acumulado.append(self.speed)

    # 2. Creación método de parar
    def parar(self):
        self.speed = 0
        self.acumulado.append(self.speed)
```

3. Cambio de velocidad

```python
carrito: Car = Car("Toyota", "Camry", 2020, 10)
carrito.freno()
print(f"{carrito.speed=}")
```

4. Distintas velocidades

```python

carrote: Car = Car("Toyota", "Camry", 2020, 10)
carrote.accelerate(20)
carrote.freno()
carrote.parar()
print(carrote.acumulado)
```
