```python
class Car:
    def __init__(self, brand:str, model:str, year:int, speed:float):
        self.brand = brand
        self.model = model
        self.year = year
        self.speed = speed

    def honk(self):
        print("Beep beep!")

    def accelerate(self, rate:float):
        self.speed += rate

    def get_speed(self):
        return self.speed
```

```python
my_car = Car("Toyota", "Camry", 2020, 0.0)
my_car.honk()
my_car.accelerate(30)
print(my_car.get_speed())
```

1. Cree una subclase llamada `ElectricCar` que herede de `Car`, mostrada en el ejemplo. Añada un atributo llamado battery-level para almacenar el nivel de batería (en porcentaje).

Implemente métodos para:
- Mostrar el nivel actual de la batería.
- Cargar la batería al 100 %.
- Sobrescribir el método honk para que emita un sonido distinto, representativo de un auto eléctrico, como Whirr beep!".

```python
TEXT_PART, SEP=1,'.'
def clean(content):
    content = str(content)
    return content.split(SEP)[TEXT_PART]
```

```python
class Electric(Car):
    def __init__(self, brand:str, model:str, year:int, speed:float, battery_level:float):
        Car.__init__(self, brand, model, year, speed)
        self.battery_level = battery_level

    def show_battery(self):
        print(clean(f'{self.battery_level=}%'))

    def full_charge(self):
        self.battery_level = 100.0

    def honk(self):
        print('Wirr beep!')
```

Ejemplos en los que defino objetos de la clase ElectricCar y los uso:

```python
car = Electric("Tesla", "Model 3", 2024, 0, 75)

car.show_battery()
# Battery level: 75%

car.full_charge()

car.show_battery()
# Battery level: 100%

car.honk()
# Whirr beep!
```

2. Cree una subclase llamada `Sport` que herede de Car. Agrege un atributo `turbo_mode` que sea `False` por defecto y un método `activate_turbo` para activarlo.
Cuando el turbo esté activado, cada vez que el carro acelere, el aumento de velocidad se multiplique por 2.
Incluya también un método deactivate-turbo para desactivar el turbo.

```python
class Sport(Car):
    def __init__(self, brand:str, model:str, year:int, speed:float):
        Car.__init__(self, brand, model, year, speed)
        self.turbo_mode = False

    def activate_turbo(self):
        self.turbo_mode = True

    def deactivate_turbo(self):
        # Considero que si se hace creación de un método `activate`
        # debe haber una forma para revertir dicha mutación en los datos.
        self.turbo_mode = False

    def accelerate(self, rate):
        new_rate = rate + rate * self.turbo_mode
        return super().accelerate(new_rate)
```

Muestro ejemplos en los que defino objetos de la clase `Sport` y los uso:

```python

sport = Sport("Ferrari", "488", 2022, 0)

sport.accelerate(20)
print(clean(f'{sport.get_speed()=}'))   # 20

sport.activate_turbo()

sport.accelerate(20)
print(clean(f'{sport.get_speed()=}'))   # 60

sport.accelerate(10)
print(clean(f'{sport.get_speed()=}'))   # 80

sport.deactivate_turbo()

sport.accelerate(10)
print(clean(f'{sport.get_speed()=}'))   # 90
```

1. Cree una subclase llamada `Truck` que herede de `Car`. Agrege un atributo `cargo_capacity` para almacenar la capacidad máxima de carga (en toneladas).

Añada un método `load_cargo` que reciba el peso de una carga y reduzca la velocidad en función del peso de la carga (por ejemplo, reducir 1 unidad de velocidad por cada 500 kg de carga). Además, incluya un método `unload_cargo` que restablezca la velocidad a la original después de descargar.

```python
TON_A_KG:float = 907.185
KG_POR_UNIDAD:float = 500.0
```

```python
class Truck(Car):
    def __init__(self, brand:str, model:str, year:int, speed:float, cargo_capacity:float):
        Car.__init__(self, brand, model, year, speed)
        self.cargo_capacity = cargo_capacity    # unidad: toneladas
        self.current_cargo = 0                  # unidad: kilogramos

    def load_cargo(self, weight: float):
        capacidad_kg = self.cargo_capacity * TON_A_KG
        if self.current_cargo + weight > capacidad_kg:
            print("Carga denegada, capacidad máxima excedida.")
            return

        print(f'Carga realizada ({weight} Kg).')
        self.current_cargo += weight
        self.speed -= weight / KG_POR_UNIDAD

    def unload_cargo(self, weight: float):
        weight = min(weight, self.current_cargo)  # no descargues más de lo que hay
        self.current_cargo -= weight
        self.speed += weight / KG_POR_UNIDAD
```

Muestro ejemplos en los que defino objetos de la clase `Truck` y los uso:

```python
# Crear un camión
truck_one = Truck("Ford", "F-150", 2000, 80, cargo_capacity=2)

truck_one.honk()                                    # Beep beep! (sin sobre-escritura)
print(clean(f'{truck_one.get_speed()=}'))           # 80

# Cargar 1000 kg    => decrementa 2 unidades de velocidad
truck_one.load_cargo(1000)
print(clean(f'{truck_one.get_speed()=}'))           # 78

# Cargar 500 kg más => decrementa 1 más
truck_one.load_cargo(500)
print(clean(f'{truck_one.get_speed()=}'))           # 77

# Descargar 500 kg  => incrementa 1 unidad
truck_one.unload_cargo(500)
print(clean(f'{truck_one.get_speed()=}'))           # 78

# Descargar el resto (1000 kg)
truck_one.unload_cargo(1000)
print(clean(f'{truck_one.get_speed()=}'))           # 80 (velocidad original es restablecida)

# Intentar exceder la capacidad máxima
truck_two = Truck("Chevrolet", "Asilveriado", 2021, 90, 1)  # solo 1 toneladita
truck_two.load_cargo(500)            # ok
truck_two.load_cargo(700)            # ko
```
