```python
class Agente:
    def __init__(self, id:str, nombre:str):
        self.id = id
        self.nombre = nombre

    def ver_id(self):
        return print(f'{self.id=}')

    def saludar(self):
        saludo = f'Hola, soy {self.nombre}'
        print(saludo)
        return saludo
```

```python
# Instancias de clase

agentito = Agente('A1', 'Juanito')
agentote = Agente('A2', 'Juanote')
agentito.ver_id()
agentote.saludar()
```

```python
class Estudiante(Agente):
    def __init__(self, id:str, nombre:str, semestre:int, carrera:str):
        Agente.__init__(self, id, nombre)
        self.semestre = semestre
        self.carrera = carrera

    def presentarse(self):
        return print(f'{self.saludar()} en la carrera de {self.carrera}')

    def aprobar_semestre(self):
        self.semestre += 1
        print(f'Cursando ahora el semestre No. {self.semestre}')

    def ver_progreso(self):
        print(f'Estás cursando {self.carrera}, el semestre No. {self.semestre}')

```

```python
estudiante_uno = Estudiante('C3', 'Pedro', 2, 'Agronomía')
estudiante_uno.presentarse()
estudiante_uno.aprobar_semestre()
estudiante_uno.ver_progreso()

estudiante_dos = Estudiante('C4', 'María', 2, 'Sistemas')
estudiante_dos.ver_id()
estudiante_dos.saludar()


```
